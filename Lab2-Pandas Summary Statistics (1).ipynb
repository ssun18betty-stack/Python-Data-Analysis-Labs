# Read the *CSV* File into a *pandas* Dataframe  

```python
import pandas as pd
df = pd.read_csv("data/titanic passenger list.csv")
```

# Part 1: How many passengers are there?
- Use `df.shape` to find how many passangers are in the dataset.

```python
print(f"number of passangers: {df.shape[0]}")
```

Part 2:

```python
df.info
```

# Part 3: How old was the youngest passenger (in months)?
- Use a pandas series object and the `min()` method to find the minimum age.

```python
youngest_age_years = df["age"].min()
youngest_age_months = youngest_age_years * 12

print(f"Youngest age years in month: {youngest_age_months}")
```

# Part 4: Identify the average price of a ticket fare.
- Use a pandas series object and the `mean()` method to find the average ticket fare.

```python
average_ticket_fare = df[f"fare"].mean()

print(f"Average ticket fare: ${average_ticket_fare:.2f}")
```

# Part 5: Calculate the combined total of all fares paid.
- Use a pandas series object and the `sum()` method to find the total of all fares paid.

```python
total_ticket_fare_paid = df[f"fare"].sum()

print(f"Total fare: ${total_ticket_fare_paid:,.2f}")
```
