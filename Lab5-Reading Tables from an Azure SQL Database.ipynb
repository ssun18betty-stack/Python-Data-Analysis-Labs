# Part 1: Show the Region, Category, Profit, and OrderDate for all sales with Profit greater than $5,900, sorted from highest to lowest Profit.

```python
df = pd.read_sql_query(
    "SELECT Region, Category, Profit, OrderDate FROM coffee_sales Where Profit > 5900 Order by Profit DESC", engine)

print(df.shape)
print("Part1: Show the Region, Category, Profit, and OrderDate for all sales with profit greater than $5900, sorted from highest to lowest profit.")
df
```

# Part 2: Show the Top 10 Sales in the East region only (sorted highest first).


```python
df_p2 = pd.read_sql_query("SELECT * FROM coffee_sales WHERE Region = 'East' ORDER BY Sales DESC", engine)

print("Part2: Show the Top 10 Sales in the East region only (sorted highest first.")
print(df_p2.shape)
df_p2.head(10)
```

# Part 3: Show the Top 10 records with Sales greater than $8,000 and Profit greater than $2,000, sorted so the highest Sales appear first.

```python
df_p3 = pd.read_sql_query("SELECT * FROM coffee_sales WHERE Sales > 8000 AND Profit > 2000 ORDER BY Sales DESC", engine)

print(df_p3.shape)
df_p3
```

# Part 4: Show the ID, Sales, and Order Date where Sales are below $1,000, sorted by Sales from lowest to highest.

```python
df_p4 = pd.read_sql_query("SELECT id, Sales, OrderDate FROM coffee_sales WHERE Sales < 1000 ORDER BY Sales ASC", engine)

print(df_p4.shape)
df_p4
```

# Part 5: Show all records from the first quarter of 2023 (January 1, 2023 to March 31, 2023), sorted by Order Date from earliest to latest.

```python
df_p5 = pd.read_sql_query(
    "SELECT * FROM coffee_sales WHERE OrderDate >= '2023-01-01' AND OrderDate <= '2023-03-31' ORDER BY OrderDate ASC", engine)

print(df_p5.shape)
df_p5.head(10)
```
