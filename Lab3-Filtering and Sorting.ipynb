#Question 1: What were the Top 10 Countries for Annual GDP Change?

```python
import pandas as pd
```

```python
df = pd.read_csv("./Data/WorldBank_GDPChange.csv")

print(df.shape)
df.head(100)
```

```python
df.dtypes
```

```python
print("Q1: What were the Top 10 Countries for Annual GDP Change?")
df1 = df.sort_values(by='GDP_Change',ascending=False)
df1 = df1.iloc[0:10]
print(df1.shape)

df1
```

#Question 2: What were the Bottom 15 Countries for Annual GDP Change?

```python
print("Q2: What were the Bottom 15 Countries for Annual GDP Change?")
df1 = df.sort_values(by='GDP_Change',ascending=True)

df1 = df1.iloc[0:15]
print(df1.shape)
df1
```
